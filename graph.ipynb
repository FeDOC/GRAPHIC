## Renew

```python
def renew():
    global otvet_col, diagnos_col, extr_col, plan_col, yellow_col, green_col, torac_col, exceptions, rest, count
    
    #count
    count = copy.deepcopy(count_base)
    
    #column_graph
    torac_col, otvet_col, yellow_col, green_col, diagnos_col, plan_col, extr_col = [copy.deepcopy(column_base) for _ in range(7)]

    #exceptions
    exceptions, rest = [copy.deepcopy(all_base) for _ in range(2)]
   
renew()  
```

## Must and Exceptions fill in

## Otvet column

```python
def create_otvet():
    choose = random.sample(otvet, k=len(otvet))
    i = 0
    for day in range(1, days_in_month + 1):
        cycle = 0
        while otvet_col[day] is None and cycle < 3:
            rand = choose[i] 
            if (day not in (exceptions[rand] + rest[rand])) and (list(otvet_col.values()).count(rand) < limits_otvet[rand]):
                otvet_col[day] = rand
                rest[rand] += [day, day + 1, day + 2, day + 3]
                count[rand] += 1
            if i == len(choose) - 1:
                i = 0
                cycle += 1
            else:
                i += 1
    return otvet_col    
```
